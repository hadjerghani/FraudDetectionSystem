```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np 

data = pd.read_csv("CleanData.csv")

X = data.drop("Class",axis=1)
y=data["Class"]

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

#Train : Logistic Regression
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=5000)
model.fit(X_train, y_train)

#Model Evaluation
y_pred = model.predict(X_test)

from sklearn.metrics import classification_report
LogisticRegressionReport = classification_report(y_test, y_pred)
print(LogisticRegressionReport)

#Confusion Matrix
#confusion Matrix
from sklearn.metrics import confusion_matrix

cm1 = confusion_matrix(y_test, y_pred)
sns.heatmap(cm1, annot=True, fmt='d')
plt.title("Logistic Regression Confusion Matrix")
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(n_estimators=100,random_state=42,class_weight='balanced')

rf_model.fit(X_train, y_train)
rf_pred = rf_model.predict(X_test)

#Model Evaluation 
from sklearn.metrics import classification_report
RandomForestReport = classification_report(y_test, rf_pred)
print(RandomForestReport)

#confusion Matrix
from sklearn.metrics import confusion_matrix

cm2 = confusion_matrix(y_test, rf_pred)
sns.heatmap(cm2, annot=True, fmt='d')
plt.title("Random Forest Confusion Matrix")
plt.show()

#Features Importance 

importance = rf_model.feature_importances_
feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': importance
})
feature_importance.sort_values(by='Importance', ascending=False).head(10)
```

```python
#Anomaly Detection 
from sklearn.ensemble import IsolationForest

iso_model = IsolationForest(n_estimators=100,contamination=0.002,random_state=42)

iso_model.fit(X_train)
#1=normal
#-1=anomaly
iso_pred = iso_model.predict(X_test)

#Convert to fraud labels 
iso_pred = [1 if x == -1 else 0 for x in iso_pred]
#1=fraud/0=normal

#Model Evaluation 
AnomalyDetectionReport = classification_report(y_test, iso_pred)
print(AnomalyDetectionReport)

cm3 = confusion_matrix(y_test, iso_pred)
sns.heatmap(cm3, annot=True, fmt='d')
plt.title("Isolation Forest Confusion Matrix")
plt.show()
```

```python
#Model Comparaison
# Logistic Regression
lr_report = classification_report(y_test, y_pred, output_dict=True)

# Random Forest
rf_report = classification_report(y_test, rf_pred, output_dict=True)

# Isolation Forest
iso_report = classification_report(y_test, iso_pred, output_dict=True)

comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest", "Isolation Forest"],
    "Fraud Recall": [
        lr_report['1']['recall'],
        rf_report['1']['recall'],
        iso_report['1']['recall']
    ],
    "Fraud Precision": [
        lr_report['1']['precision'],
        rf_report['1']['precision'],
        iso_report['1']['precision']
    ],
    "Fraud F1-score": [
        lr_report['1']['f1-score'],
        rf_report['1']['f1-score'],
        iso_report['1']['f1-score']
    ]
})

comparison
```

```python
#ROC-AUC
from sklearn.metrics import roc_curve, roc_auc_score
lr_probs = model.predict_proba(X_test)[:,1]
rf_probs = rf_model.predict_proba(X_test)[:,1]
iso_scores = iso_model.decision_function(X_test)
iso_probs = -iso_scores#highr anomaly=higher fraudrisk

#ROC curves
lr_fpr, lr_tpr, _ = roc_curve(y_test, lr_probs)
rf_fpr, rf_tpr, _ = roc_curve(y_test, rf_probs)
iso_fpr, iso_tpr, _ = roc_curve(y_test, iso_probs)

#AUC scores
lr_auc = roc_auc_score(y_test, lr_probs)
rf_auc = roc_auc_score(y_test, rf_probs)
iso_auc = roc_auc_score(y_test, iso_probs)

plt.figure(figsize=(8,6))

plt.plot(lr_fpr, lr_tpr, label=f"Logistic Regression (AUC = {lr_auc:.3f})")
plt.plot(rf_fpr, rf_tpr, label=f"Random Forest (AUC = {rf_auc:.3f})")
plt.plot(iso_fpr, iso_tpr, label=f"Isolation Forest (AUC = {iso_auc:.3f})")

plt.plot([0,1],[0,1],'k--')  # random model line

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison for Fraud Detection")

plt.legend()
plt.show()
```
